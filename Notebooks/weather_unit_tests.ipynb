```python

def calculate_temperature_band(temp):

    if temp < 15:
        return "Cold"

    elif temp <= 30:
        return "Normal"

    elif temp <= 40:
        return "Hot"

    return "Extreme"


def calculate_severity_score(temp, rainfall, wind):

    score = 0

    if temp > 40:
        score += 3

    if rainfall > 30:
        score += 2

    if wind > 40:
        score += 2

    return score


def get_alert_type(severity, weather, wind, rainfall):

    if severity >= 4:
        return "heatwave"

    if weather:

        weather = weather.lower()

        if "storm" in weather:
            return "storm"

        if "thunder" in weather:
            return "storm"

    if wind > 50:
        return "high_wind"

    if rainfall > 50:
        return "heavy_rain"

    return None
```

```python

# TEMPERATURE BAND TESTS

assert calculate_temperature_band(10) == "Cold"
assert calculate_temperature_band(25) == "Normal"
assert calculate_temperature_band(35) == "Hot"
assert calculate_temperature_band(45) == "Extreme"

print(" temperature_band tests passed")

# SEVERITY SCORE TESTS

assert calculate_severity_score(45, 50, 60) == 7
assert calculate_severity_score(42, 0, 0) == 3
assert calculate_severity_score(20, 0, 0) == 0

print(" severity_score tests passed")

# ALERT TYPE TESTS

assert get_alert_type(5,"clear sky",0,0) == "heatwave"
assert get_alert_type(0,"thunderstorm",0,0) == "storm"
assert get_alert_type(0,"cloudy", 60,0) == "high_wind"
assert get_alert_type(0,"cloudy",0,55) == "heavy_rain"

print(" alert_type tests passed")
print("ALL UNIT TESTS PASSED")
```
